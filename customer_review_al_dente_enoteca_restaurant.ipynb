```python
from langchain_community.llms import Replicate
import os
from google.colab import userdata
# Set the API token
api_token = userdata.get('api_token')
os.environ["REPLICATE_API_TOKEN"] = api_token
# Model setup
model = "ibm-granite/granite-3.2-8b-instruct"
output = Replicate(
model=model,
replicate_api_token=api_token,
)
```

```python
# Define the customer reviews
customer_reviews = [
"Excellent service and very welcoming staff. The scallop entree was fresh and delicious, the sauce paired beautifully with the crispy fried shallots.The duck breast was juicy and tender with rich aroma from the duck oil and a nicely crisped skin, very satisfying. The only disappointment was the Wagyu M9. We ordered it medium, but it was served almost well done, which ruined the expected tenderness and texture. It ended up dry and tough to chew. The portion was little small, still had room for dessert after meal!.",
"This is our second time dining at Al Dente, and it had been amazing both times! The service is always welcoming and warm. The food is a twist on Italian classics. The Casio e pepe hasn’t changed since we last ate in the restaurant which was one of the reasons we wanted to come back. It never misses a beat. The lasagna was cooked to perfection, just like mama used to make it ;) The Kingfish and crab ravioli was my personal favorite, so well balanced with so delicate ingredients. Thank you for celebrating my birthday with us and we will definitely see each other again!",
"Very nice service, I ordered the rabbit rugu, braised boar, scallops and the salted butter. Rabbit Rugu was great, something out of ordinary flavour. The salted butter I didn't find particularly good but my kids loved it with the sourdough bread. The restaurant is spacious, the table and plates were cleaned up on each round of the serving. A place where you can relax and enjoy your night.",
"We were initially given sparkling water instead of the tap water we requested. The sparkling water was then taken away without explanation after it was poured into our cups... they were rude about how they go with it. When our wine was served, only two glasses were brought to the table, despite there being four of us drinking and had to be reminded twice to bring it over. the staff member mistakenly ordered a double portion of pasta. When we brought this to the managers attention, she was condescending and sarcastic. She implied we had ordered the extra dish and then acted as if she was doing us a favor by removing it from the bill, even though we didnt eat it and it was their mistake. When we were ready to pay, the staff at the cashier were engaged in a loud and distracting conversation, ignoring our attempts to get their attention. The food itself was average and certainly not worth the high prices charged. Overall, we found the experience to be very disappointing and would not recommend this restaurant."
]

# Refine the prompt to include reviews
reviews_text = "\n".join([f"Review {i+1}: {review}" for i, review
in enumerate(customer_reviews)])
prompt = f"""
Classify all reviews into three categories Positive, Negative, or Mixed:
{reviews_text}
"""
# Invoke the model with the example prompt
response = output.invoke(prompt)
# Print the response
print("Granite Model Response:\n")
print(response)
```

```python
# Define refined prompt
refined_prompt = f"""
Classify these reviews as positive, negative, or mixed, and tag
relevant categories (food quality, customer service, or
atmosphere):
{reviews_text}
"""
# Invoke the model with the example prompt
response = output.invoke(refined_prompt)
# Print the response
print("Granite Model Refined Response:\n")
print(response)
```

```python
# Define the prompt to complete the task in 2 steps
multitask_prompt = f"""
Complete the task in 2 steps.
Step 1: Classify these reviews as positive, negative, or mixed.
Step 2: For each review, identify relevant categories: food quality, service, or atmosphere.
{reviews_text}
"""
response = output.invoke(multitask_prompt)
print("Granite Model Response:\n")
print(response)
```

```python
# Define the example to guide the model
formatted_prompt = f"""
Classify these reviews as Positive, Negative, or Mixed, and tag
relevant categories. Use this format:
- Sentiment: [Sentiment]
- Categories: [Categories].

{reviews_text}
"""
# Invoke the model with prompt
response = output.invoke(formatted_prompt)
# Print the response
print("Granite Model Formatted Response:\n")
print(response)
```
